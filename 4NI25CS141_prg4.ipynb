```python
#importing libraries
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv("/content/melb_data.csv")
print("Original Dataset:")
print(df.head())
print("\nShape of the dataset:")
print(df.shape)
print("\nColumns of the dataset:")
print(df.columns)
print("\nDataset Information")
print(df.info())
print("\nStatistical Summary:")
print(df.describe())
```

```python
#check for missing values
print("\nMissing Values:")
print(df.isnull().sum())

#------------------------------------------------------------------------------
#MEAN IMPUTATION
#------------------------------------------------------------------------------

#Create a copy of the dataset
df_mean = df.copy()

#calculate the mean of Building area
mean_building_area = df_mean["BuildingArea"].mean()

#Reaplace missing BuildingArea values with the mean
df_mean["BuildingArea"] = df_mean["BuildingArea"].fillna(mean_building_area)

print("\nMean Building Area:", mean_building_area)
print("\nMissing BuildingArea Values After Mean Imputation:")
print(df_mean["BuildingArea"].isnull().sum())

df_median = df.copy()

#calculate the median of Building area
median_building_area = df_median["BuildingArea"].median()

#Reaplace missing BuildingArea values with the median
df_median["BuildingArea"] = df_median["BuildingArea"].fillna(median_building_area)

print("\nMedian Building Area:", median_building_area)
print("\nMissing BuildingArea Values After Median Imputation:")
print(df_median["BuildingArea"].isnull().sum())
```
