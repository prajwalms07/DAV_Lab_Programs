```python
# CHECK DUPLICATE ROWS
print("\nNumber of duplicate rows:")
print(df.duplicated().sum())
```

```python
# CHECK DUPLICATE STUDENT IDs
print("\nNumber of duplicate student IDs:")
print(df['student_id'].duplicated().sum())
```

```python
# DISPLAY GRADE COUNTS
print("\nNumber of students in each grade:")
print(df['grade'].value_counts())
```

```python
# DATA WRANGLING
# CREATE A NEW DATAFRAME FOR MERGING
grade_info = pd.DataFrame({
    'grade': ['A', 'B', 'C', 'D', 'F'],
    'grade_description': [
        'Excellent',
        'Good',
        'Average',
        'Below Average',
        'Fail'
    ]
})

print("\nGrade Information DataFrame:")
print(grade_info)
```

```python
# MERGING TWO DATAFRAMES
merged = pd.merge(
    df,
    grade_info,
    on='grade',
    how='left'
)
print("\nMerged DataFrame:")
print(merged.head())
```

```python
# CREATE DATAFRAME FOR JOIN
grade_details = pd.DataFrame({
    'description': [
        'Excellent',
        'Good',
        'Average',
        'Below Average',
        'Fail'
    ],
    'performance_level': [
        'Very High',
        'High',
        'Medium',
        'Low',
        'Very Low'
    ]
    }, index=['A', 'B', 'C', 'D', 'F'])

# COUNT STUDENTS BY GRADE
grade_counts = df['grade'].value_counts().to_frame('student_count')

# JOIN TWO DATAFRAMES
joined = grade_counts.join(grade_details)
print("\nJoined DataFrame:")
print(joined)
```

```python
# CONCATENATION
# SPLIT DATAFRAME INTO TWO PARTS
df1 = df.iloc[:100]
df2 = df.iloc[100:200]
combined = pd.concat([df1, df2])

print("\nConcatenated DataFrame:")
print(combined.head())

print("\nShape after concatenation:")
print(combined.shape)
```
