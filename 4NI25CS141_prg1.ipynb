```python
import numpy as np
import pandas as pd
print("\nLibraries Imported")
```

```python
# NumPy array operations
arr = np.array([10, 20, 30, 40, 50])
print('Array:', arr)
print('Mean:', arr.mean())
print('Reshaped 2D:\n', arr.reshape(5,1))
print('Sliced (index 1 to 3):', arr[1:4])
```

```python
# --- Create a DataFrame ---
data = {'Name': ['Asha','Ravi','Meera','Karan'],
        'Marks': [88, 72, 91, 65],
        'Branch': ['CSE','ECE','CSE','ISE']}
df = pd.DataFrame(data)
print(df)
```

```python
# --- Indexing and slicing ---
print('\nFirst 2 rows:\n', df.iloc[0:2])
print('\nMarks column:\n', df['Marks'])
print('\nRows where Marks > 70:\n', df[df['Marks'] > 70])
```
